```python
import csv
with open('trainingdata.csv', 'r') as file:
    data = list(csv.reader(file))
print("Training Data:\n")
for row in data:
    print(row)
num_attributes = len(data[0]) - 1

hypothesis = ['0'] * num_attributes

for row in data:
    if row[-1].strip().lower() == "yes":
        for i in range(num_attributes):
            if hypothesis[i] == '0':
                hypothesis[i] = row[i]
            elif hypothesis[i] != row[i]:
                hypothesis[i] = '?'
print("\nFinal Hypothesis:")
print(hypothesis)
```

```python
import csv
def initialize_s(num_attr):
    return ['0'] * num_attr
def initialize_g(num_attr):
    return [['?' for _ in range(num_attr)] for _ in range(num_attr)]
with open('trainingdata.csv', 'r') as file:
    data = list(csv.reader(file))
header = data[0]
data = data[1:]

num_attr = len(data[0]) - 1
S = initialize_s(num_attr)
G = initialize_g(num_attr)

print("Initial Specific Hypothesis (S):")
print(S)

print("\nInitial General Hypothesis (G):")
for g in G:
    print(g)
for example in data:
    attributes = example[:-1]
    target = example[-1].strip().lower()

    if target == "yes":
        # Update Specific Hypothesis
        for i in range(num_attr):
            if S[i] == '0':
                S[i] = attributes[i]
            elif S[i] != attributes[i]:
                S[i] = '?'
                G[i][i] = '?'

    elif target == "no":
        # Update General Hypothesis
        for i in range(num_attr):
            if S[i] != attributes[i]:
                G[i][i] = S[i]
            else:
                G[i][i] = '?'

    print("\nTraining Example:", example)
    print("Specific Hypothesis (S):", S)
    print("General Hypothesis (G):")
    for g in G:
        print(g)
G = [g for g in G if g != ['?' for _ in range(num_attr)]]

print("\nFinal Specific Hypothesis:")
print(S)

print("\nFinal General Hypothesis:")
for g in G:
    print(g)
```
