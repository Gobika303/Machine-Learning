```python
import numpy as np

a = np.array([1, 2, 3, 4, 5])
print(a)

print(a.shape)
print(a.size)
print(a.dtype)

b = np.arange(1, 11)
print(b)

c = np.linspace(1, 10, 5)
print(c)

z = np.zeros((3, 3))
o = np.ones((3, 3))
i = np.identity(3)

print(z)
print(o)
print(i)

r = np.random.rand(3, 3)
print(r)

x = np.array([1, 2, 3, 4, 5, 6])

print(x.sum())
print(x.mean())
print(x.max())
print(x.min())
print(x.std())
print(x.var())

m = np.array([[1, 2], [3, 4]])
n = np.array([[5, 6], [7, 8]])

print(np.add(m, n))
print(np.subtract(m, n))
print(np.multiply(m, n))
print(np.matmul(m, n))

print(m.reshape(4, 1))
print(m.ravel())
```

```python
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)
y = x ** 2

plt.plot(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Square Function")
plt.grid(True)
plt.show()
```

```python
# Keras
# -----------------------------
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model = Sequential()

model.add(Dense(16, input_dim=4, activation="relu"))
model.add(Dense(8, activation="relu"))
model.add(Dense(3, activation="softmax"))

model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```
