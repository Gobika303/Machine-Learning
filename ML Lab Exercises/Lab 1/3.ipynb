```python
import math

num = 25.8

print("Number =", num)
print("Floor =", math.floor(num))
print("Ceil =", math.ceil(num))
print("Square Root of 64 =", math.sqrt(64))
print("Integer Square Root of 25 =", math.isqrt(25))
print("GCD of 24 and 36 =", math.gcd(24, 36))
```

```python
import numpy as np

arr = np.array([[10, 20, 30],
                [40, 50, 60]])

print("Array:")
print(arr)

print("Number of Dimensions:", arr.ndim)
print("Shape:", arr.shape)
print("Size:", arr.size)

print("Sum:", arr.sum())
print("Mean:", arr.mean())

print("Sorted Array:")
print(np.sort(arr))

print("Sine Values:")
print(np.sin(arr))
```

```python
import numpy as np
from scipy import linalg

A = np.array([[2, 3],
              [1, 4]])

print("Matrix:")
print(A)

print("Determinant =", linalg.det(A))

eigen_values, eigen_vectors = linalg.eig(A)

print("Eigen Values:")
print(eigen_values)

print("Eigen Vectors:")
print(eigen_vectors)
```
